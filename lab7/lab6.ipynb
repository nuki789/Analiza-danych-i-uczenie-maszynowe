```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def regression_coef(x, y):
    
    licznik = (x-np.mean(x)) * (y-np.mean(y))
    mianownik = (x-np.mean(x))**2

    a = licznik.sum() / mianownik.sum()
    b = np.mean(y) - a*np.mean(x)

    return a, b
```

```python
def regression_linear_plot(x, y, coef):
    plt.figure(figsize=(10, 6))
    plt.scatter(x, y, color='r', marker='o', s=30)

    y_pred = coef[0] * x + coef[1]
    plt.plot(x, y_pred, color='b')

    plt.title('Regresja liniowa')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid('on')
    plt.show()
```

```python
np.random.seed(0)
n = 100
x = 10 * np.random.rand(n) - 5
y = 5 * x + 10 + 20 * np.random.randn(n)
```

```python
a, b = regression_coef(x, y)
coef = (a, b)
regression_linear_plot(x, y, coef)
```
