```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
def regression_coef(X, y):
    x_mean = np.mean(X)
    y_mean = np.mean(y)
    
    xy_cov = (X-x_mean)*(y-y_mean)
    x_var = (X-x_mean)**2
    
    a = xy_cov.sum() / x_var.sum()
    b = y_mean - a*x_mean
    
    return a, b
```

```python
class RegLin:
    def __init__(self):
        self.coef_ = [0, 0]

    def fit(self, X, y):
        x_mean = np.mean(X)
        y_mean = np.mean(y)
        xy_cov = (X-x_mean) * (y-y_mean)
        x_var = (X-x_mean)**2
        a = xy_cov.sum() / x_var.sum()
        b = y_mean - a*x_mean
        self.coef_[0] = a
        self.coef_[1] = b
    def predict(self, X):
        return self.coef_[0] * X + self.coef_[1]
```

```python
model = RegLin()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

model.fit(x, y)
```

```python
model.predict(np.array([1, 2, 3, 4, 5]))
```

```python
def rysuj(X, y, model):
    plt.figure(figsize=(10, 6))
    plt.scatter(X, y, color = 'r', marker = 'o', s=30)
    y_pred = model.coef_[0]*np.array(X) + model.coef_[1]
    plt.plot(X, y_pred, '-b') 
    plt.title('Regresja liniowa')
    plt.xlabel('X')
    plt.ylabel('y')
    plt.grid('on')
    plt.show()
```

```python
rysuj(x, y, model)
```
