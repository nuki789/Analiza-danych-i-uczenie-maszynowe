```python
dane = []

for i in range(11):
    dane.append(i)
    
plik = open("dane.txt", "w")
for i in range(0, len(dane)):
    plik.write(str(dane[i]) + "\n")

plik.close()
```

```python
from string import ascii_lowercase as alfabet
plik = open("dane.txt", "a")

for litera in alfabet:
    plik.write(litera + "\n")

plik.close()
```
