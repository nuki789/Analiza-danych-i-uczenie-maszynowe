```python
import math as mt

a = 1
b = 1
c = 1

plik = open("tablica.csv", "w")
plik.write(f"i,x,sin,cos,tg,ctg,a*x**2+b*x+c\n")
for i in range(1, 640):
    x = 0.01 * i
    sin = a * mt.sin(b * x)
    cos = a * mt.cos(b * x)
    tg = a * mt.tan(b * x)
    ctg =a * (1/tg)
    r=a*x**2 + b*x + c
    plik.write(f"{i}; {x}; {sin}; {cos}; {tg}; {ctg}; {r}\n")
plik.close()
```
